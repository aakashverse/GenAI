##### 4. Multi-Query Retriever
###### sometimes a single query might not capture all the information that are phrased in your documents. MQ retriever tries to remove the ambiguity in query and search the relevant results without duplication

```python
from langchain_community.vectorstores import Chroma
from langchain_google_genai import GoogleGenerativeAIEmbeddings
from langchain_groq import ChatGroq
from langchain_core.retrievers import MultiQueryRetriever
from langchain_core.documents import Document 
```

```python
embedding_model = GoogleGenerativeAIEmbeddings(model="gemini-embedding-2-preview")
```

```python
docs = [
    Document(page_content="High blood sugar is a common sign of diabetes.", metadata={"source": "H1"}),
    Document(page_content="Frequent urination may indicate diabetes.", metadata={"source": "H2"}),
    Document(page_content="Excessive thirst is associated with diabetes.", metadata={"source": "H3"}),
    Document(page_content="Type 2 diabetes affects insulin regulation.", metadata={"source": "H4"}),
    Document(page_content="Fatigue can occur in people with diabetes.", metadata={"source": "H5"}),
]
```

```python
vector_store = Chroma.from_documents(
    documents=docs,
    embedding=embedding_model
)
```

```python
similarity_retriever = vector_store.as_retriever(
    search_type="similarity",
    search_kwargs={"k":3}
)
```

```python
multiquery_retriever = MultiQueryRetriever.from_llm(
    retriever=vector_store.as_retriever(search_kwargs={"k":3}),
    llm=ChatGroq(model="openai/gpt-oss-20b")
)
```

```python
query = "Why am I always thirsty and tired?"
similarity_result = similarity_retriever.invoke(query)
multiquery_result = multiquery_retriever.invoke(query)
```

```python
for i,doc in enumerate(similarity_result):
    print(f"...Result {i+1}..."),
    print(doc.page_content)

    print("*"*120)

for i,doc in enumerate(multiquery_result):
    print(f"...Result {i+1}..."),
    print(doc.page_content)
```
