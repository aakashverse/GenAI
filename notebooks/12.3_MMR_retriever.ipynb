##### 3. Maximal Marginal Relevance (MMR) - (diverse but relevant info)
###### MMR is an information retrieval algorithm designed to reduce redundancy in the retrieved results while maintaining high relevance to the query.

```python
from langchain_community.vectorstores import Chroma
from langchain_google_genai import GoogleGenerativeAIEmbeddings
from langchain_core.documents import Document
```

```python
documents = [
    Document(page_content="LangChain helps developers build LLM-powered applications efficiently."),
    Document(page_content="LangChain is used to build LLM based applications."),
    Document(page_content="Retrieval-Augmented Generation (RAG) combines LLMs with external knowledge sources."),
    Document(page_content="Vector databases store embeddings for semantic search and similarity matching."),
    Document(page_content="Langchain supports Chroma, FAISS, Pinecone, and more."),
    Document(page_content="Agents in LangChain can use tools and reasoning to solve complex tasks.")
]
```

```python
embedding_model = GoogleGenerativeAIEmbeddings(model="gemini-embedding-2-preview")
```

```python
vector_store = Chroma.from_documents(
    documents=documents,
    embedding=embedding_model
)
```

```python
retriever = vector_store.as_retriever(
    search_type="mmr",
    search_kwargs={"k":3, "lambda_mult": 0.5}   # lambda_mult -> diversity balance
)
```

```python
query = "what is langchain"
result = retriever.invoke(query)
```

```python
for i,doc in enumerate(result):
    print(f"\n...Result {i+1}...")
    print(doc.page_content)
```
