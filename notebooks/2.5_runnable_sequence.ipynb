#### Runnable Primitive


##### 1) Runnable Sequence- RS is a sequential chain of runnables that executes each step one after another, passing the output of one step as the input to the next. It is useful when u need to compose multiple runnables together in a structured workflow

```python
from langchain_huggingface import ChatHuggingFace, HuggingFaceEndpoint
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnableSequence
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
prompt1 = PromptTemplate(
    template='Give a joke on {topic}',
    input_variables=['topic']
)
```

```python
prompt2 = PromptTemplate(
    template='Explain the joke {text}',
    input_variables=['text']
)
```

```python
llm = HuggingFaceEndpoint(
    repo_id="meta-llama/Llama-3.1-8B-Instruct",
    task="text-generation"
)

model = ChatHuggingFace(llm=llm)
```

```python
parser = StrOutputParser()
```

```python
chain = RunnableSequence(prompt1, model, parser, prompt2, model, parser)
```

```python
print(chain.invoke({'topic':'AI'}))
```
