##### RunnableBranch - conditional chain/ control component to conditionally route input data to diff chains or runnables

```python
from langchain_huggingface import ChatHuggingFace, HuggingFaceEndpoint
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnableSequence, RunnablePassthrough, RunnableBranch
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
prompt1 = PromptTemplate(
    template='Generate a detailed report on {topic}',
    input_variables=['topic']
)
```

```python
prompt2 = PromptTemplate(
    template='Summarize the given text- {text}',
    input_variables=['text']
)
```

```python
llm = HuggingFaceEndpoint(
    repo_id="meta-llama/Llama-3.1-8B-Instruct",
    task="text-generation"
)

model = ChatHuggingFace(llm=llm)
```

```python
parser = StrOutputParser()
```

```python
report_gen_chain = RunnableSequence(prompt1, model, parser)
```

```python
branch_chain = RunnableBranch(
    (lambda x: len(x.split())>300, RunnableSequence(prompt2, model, parser)),
    RunnablePassthrough()
)
```

```python
final_chain = RunnableSequence(report_gen_chain, branch_chain)
```

```python
print(final_chain.invoke({'topic':'AGI'}))
```

##### LCEL (Langchain Expression Language)- RunnableSequence very frequent (almost everywhere) -> a declarative way of defining RunnableSequence using pipe (|) operator
##### like, report_gen_chain = prompt1 | model | parser
