#### Solution: Structured Output (helps in Data Extraction, API building & Agents)
##### structured output (along with Pydantic) refers to the practice of having language models return responses in a well defined data format  (like JSON), rather than free-from text. This makes the model output easier to parse and work with programatically. (Schema can be enforced for the LLM output)

```python
from pydantic import BaseModel, Field # pydantic - data validation & data parsing library ensures the data you work with is correct, structured & type-safe
from typing import List
```

```python
class ResponseSchema(BaseModel):
    name: str = Field(description="Full Name")     # adding field validation
    email: str = Field(description="Email Address")
    age: int = Field(description="Age")

structured_llm = llm.with_structured_output(ResponseSchema)
```

##### structured output (by 3 ways) => 1) TypeDict, 2) Pydantic, 3) Json schema

```python
res = structured_llm.invoke(f"Please give me only name, email and age from this text: {text}")
res
```

```python
class Movies(BaseModel):
    title: str = Field(description="Movie Title")
    year: int = Field(description="Movie Released Year") 

class AllMovies(BaseModel):
    movies: List[Movies]

movie_llm = llm.with_structured_output(AllMovies)
```

##### 'with_structured_output' have two methods of data formating (ways) => 1) Json Mode  2)  function calling (mostly used in Agents)

```python
res = movie_llm.invoke("Give three Sci-Fi movie.")
```
