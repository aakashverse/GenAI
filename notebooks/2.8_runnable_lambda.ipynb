##### RunnableLambda- is a runnable primitive that allows u to apply custom Python function within an AI pipeline, It acts a middleware btw different AI components enabling preprocessing, transformation, API calls, filtering & post-processing.

```python
from langchain_huggingface import ChatHuggingFace, HuggingFaceEndpoint
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnableSequence, RunnableParallel, RunnablePassthrough, RunnableLambda
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
prompt = PromptTemplate(
    template='Give a joke on {topic}',
    input_variables=['topic']
)
```

```python
llm = HuggingFaceEndpoint(
    repo_id="meta-llama/Llama-3.1-8B-Instruct",
    task="text-generation"
)

model = ChatHuggingFace(llm=llm)
```

```python
parser = StrOutputParser()
```

```python
joke_gen_chain = RunnableSequence(prompt, model, parser)
```

```python
parallel_chain = RunnableParallel({
    'joke': RunnablePassthrough(),
    'count_words': RunnableLambda(lambda x: len(x.split()))
})
```

```python
final_chain = RunnableSequence(joke_gen_chain, parallel_chain)
```

```python
result = final_chain.invoke({'topic':'AI'})
```
