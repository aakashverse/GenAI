##### RunnablePassthrough - a special runnable primitive that simply returns the input as output without modifying it

```python
from langchain_huggingface import ChatHuggingFace, HuggingFaceEndpoint
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnableSequence, RunnableParallel, RunnablePassthrough
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
prompt1 = PromptTemplate(
    template='Generate a joke on {topic}',
    input_variables=['topic']
)
```

```python
prompt2 = PromptTemplate(
    template='Explain the following text - {text}',
    input_variables=['text']
)
```

```python
llm = HuggingFaceEndpoint(
    repo_id="meta-llama/Llama-3.1-8B-Instruct",
    task="text-generation"
)

model = ChatHuggingFace(llm=llm)
```

```python
parser = StrOutputParser()
```

```python
joke_generator_chain =  RunnableSequence(prompt1, model, parser)
```

```python
parallel_chain = RunnableParallel({
    'joke': RunnablePassthrough(),
    'explanation': RunnableSequence(prompt2, model, parser)
})
```

```python
chain = RunnableSequence(joke_generator_chain, parallel_chain)
```

```python
print(chain.invoke({'topic':'cricket'}))
```
