```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_huggingface import ChatHuggingFace, HuggingFaceEndpoint
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnableParallel
```

```python
prompt1 = PromptTemplate(
    template='Generate short & simple notes on the given topic \n {text}',
    input_variables=['text']
)
```

```python
prompt2 = PromptTemplate(
    template='Generate 5 short questions answers using the following text \n {text}',
    input_variables=['text']
)
```

```python
prompt3 = PromptTemplate(
    template='Merge the provided notes & quiz into a single document \n notes -> {notes} and quiz -> {quiz}',
    input_variables=['notes', 'quiz']
)
```

```python
llm = HuggingFaceEndpoint(
    repo_id="meta-llama/Llama-3.1-8B-Instruct",
    task="text-generation"
)

model = ChatHuggingFace(llm=llm)
```

```python
parser = StrOutputParser()
```

```python
parallel_chain = RunnableParallel({
    'notes': prompt1 | model | parser,
    'quiz' : prompt2 | model | parser 
})
```

```python
merge_chain = prompt3 | model | parser
```

```python
chain = parallel_chain | merge_chain
```

```python
text = """
1. Introduction

Quantum computing is an emerging area of computation that uses principles of quantum mechanics—the fundamental theory describing nature at atomic and subatomic scales—to process information in ways that classical computers cannot easily replicate.

Unlike classical computers that use bits (0 or 1), quantum computers use quantum bits or qubits, which can exist in multiple states simultaneously due to a property called superposition.

The field is closely tied to the broader Quantum Computing domain and has the potential to transform industries such as cryptography, drug discovery, optimization, and artificial intelligence.

2. Key Principles
2.1 Qubits

A qubit is the basic unit of quantum information. Unlike a classical bit:

Classical bit: either 0 or 1
Qubit: can be 0, 1, or both at the same time (superposition)
2.2 Superposition

This allows quantum systems to represent many possibilities at once, enabling massive parallel computation.

2.3 Entanglement

When qubits become entangled, the state of one qubit instantly relates to another, even if they are far apart. This property enables extremely powerful correlations in computation.

2.4 Quantum Interference

Quantum states can reinforce or cancel each other, helping algorithms amplify correct answers and suppress wrong ones.

3. How Quantum Computers Work

Quantum computers operate using:

Quantum circuits (like logic circuits in classical computing)
Quantum gates (operations applied to qubits)
Measurement (collapsing quantum states into classical results)

The process typically involves preparing qubits, applying quantum gates, and measuring the output.

4. Major Quantum Algorithms
4.1 Shor’s Algorithm
Used for factoring large numbers efficiently
Threatens classical encryption systems like RSA
4.2 Grover’s Algorithm
Speeds up searching unsorted databases
Provides quadratic speedup over classical search
4.3 Quantum Simulation Algorithms
Simulate molecules and materials
Extremely useful for chemistry and pharmaceutical research
5. Applications
5.1 Cryptography

Quantum computers could break many current encryption methods, pushing development of post-quantum cryptography.

5.2 Drug Discovery

Simulating molecules at quantum level may accelerate the design of new medicines.

5.3 Optimization Problems

Useful in logistics, supply chain, traffic routing, and financial modeling.

5.4 Artificial Intelligence

Quantum machine learning may improve pattern recognition and training efficiency.

6. Current Challenges
6.1 Decoherence

Quantum states are extremely fragile and easily disrupted by environmental noise.

6.2 Error Rates

Quantum operations are prone to errors, requiring complex error correction methods.

6.3 Scalability

Building large-scale quantum systems with many stable qubits is still difficult.

6.4 Cost and Infrastructure

Quantum computers require extremely low temperatures and specialized environments.

7. Leading Technologies and Companies

Some major players advancing quantum computing include:

IBM Quantum
Google Quantum AI
Microsoft Azure Quantum
Intel Quantum Research
Rigetti Computing

These organizations are developing both hardware (quantum processors) and software ecosystems.

8. Future Outlook

Quantum computing is still in the Noisy Intermediate-Scale Quantum (NISQ) era, meaning current machines are powerful but not yet fault-tolerant.

In the future, fully error-corrected quantum computers could:

Break current cryptographic systems
Revolutionize drug discovery
Solve complex optimization problems beyond classical reach

However, it is unlikely to fully replace classical computing; instead, it will complement it for specialized tasks.

9. Conclusion

Quantum computing represents a paradigm shift in computation, leveraging the laws of quantum physics to solve certain problems dramatically faster than classical computers. While still in its early stages, its potential impact on science, technology, and industry is profound.
"""
```

```python

result = chain.invoke({'text':text})
```
