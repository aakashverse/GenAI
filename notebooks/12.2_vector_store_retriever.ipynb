##### 2. Vector Store Retriever
###### it is the most common type of retriever that lets u fetch docs from a vector store based on semantic similarity using vector embeddings

```python
from langchain_community.vectorstores import Chroma
from langchain_google_genai import GoogleGenerativeAIEmbeddings
from langchain_core.documents import Document
```

```python
documents = [
    Document(page_content="LangChain helps developers build LLM-powered applications efficiently."),
    Document(page_content="Retrieval-Augmented Generation (RAG) combines LLMs with external knowledge sources."),
    Document(page_content="Vector databases store embeddings for semantic search and similarity matching."),
    Document(page_content="Prompt engineering improves the quality and reliability of LLM responses."),
    Document(page_content="Agents in LangChain can use tools and reasoning to solve complex tasks.")
]
```

```python
embedding_model=GoogleGenerativeAIEmbeddings(model = "gemini-embedding-2-preview")

vector_store=Chroma.from_documents(
    documents=documents,
    embedding=embedding_model,
    collection_name="my_collection"
)
```

```python
retriever = vector_store.as_retriever(kwargs={"k":2})
```

```python
query = "Why RAG is used for?"
result = retriever.invoke(query)
```

```python
for i,doc in enumerate(result):
    print(f"\n...Result {i+1}...")
    print(doc.page_content)
```
