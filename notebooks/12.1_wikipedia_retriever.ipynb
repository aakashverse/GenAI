#### Retrievers - A retriever is a langchain component that fetches relevant documents from a data source in response to a user's query

##### Types - [1] on basis of data source/flow

##### 1. Wikipedia retriever
###### A wikipedia retriever that queries the wikipedia API to fetch relevant content for a given query

```python
from langchain_community.retrievers import WikipediaRetriever
```

```python
retriever = WikipediaRetriever(top_k_results=2, lang='en')
```

```python
query="Extraterrestrial life"
docs = retriever.invoke(query)
```

```python
for i, doc  in enumerate(docs):
    print(f"\n...Result {i+1}...")
    print(f"Content:\n{doc.page_content}...")
```
