##### Runnable Parallel - is a runnable primitive that allows multiple runnables to execute in parallel, each runnable receives the same input & processes it independantly, producing a dictionary of output

```python
from langchain_huggingface import ChatHuggingFace, HuggingFaceEndpoint
from langchain_core.output_parsers import StrOutputParser
from langchain_core.prompts import PromptTemplate
from langchain_core.runnables import RunnableParallel, RunnableSequence
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
prompt1 = PromptTemplate(
    template='Generate a tweet on {topic}',
    input_variables=['topic']
)
```

```python
prompt2 = PromptTemplate(
    template='Generate a linkedin post on {topic}',
    input_variables=['topic']
)
```

```python
llm = HuggingFaceEndpoint(
    repo_id="meta-llama/Llama-3.1-8B-Instruct",
    task="text-generation"
)

model = ChatHuggingFace(llm=llm)
```

```python
parser = StrOutputParser()
```

```python
parallel_chain = RunnableParallel({
    'tweet': RunnableSequence(prompt1, model, parser),
    'post' : RunnableSequence(prompt2, model, parser)
})
```

```python
result = parallel_chain.invoke({'topic':'AI'})
```

```python
print(result['tweet'])
print(result['post'])
```
